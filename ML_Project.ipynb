```python
import cv2
import pytesseract
from PIL import Image
import os
import pandas as pd

# Set Tesseract-OCR path (only for Windows users)
# Example: r"C:\Program Files\Tesseract-OCR\tesseract.exe"
pytesseract.pytesseract.tesseract_cmd = r"/usr/bin/tesseract"  # Adjust if needed

def extract_text_from_image(image_path):
    """
    Extracts text from a single image using Tesseract OCR.
    """
    # Load the image
    image = cv2.imread(image_path)

    # Convert image to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Apply thresholding (Otsu’s method) for better OCR
    gray = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]

    # Perform OCR
    extracted_text = pytesseract.image_to_string(gray)

    return extracted_text.strip()

def process_images_in_folder(folder_path, output_csv="extracted_texts.csv"):
    """
    Processes all images in a given folder, extracts text, and saves results to a CSV file.
    """
    results = []
    image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".tiff")  # Supported formats

    for filename in os.listdir(folder_path):
        if filename.lower().endswith(image_extensions):  # Check if it's an image
            image_path = os.path.join(folder_path, filename)
            extracted_text = extract_text_from_image(image_path)

            results.append({"Filename": filename, "Extracted_Text": extracted_text})

            print(f"Processed: {filename}")

    # Save results to CSV
    df = pd.DataFrame(results)
    df.to_csv(output_csv, index=False, encoding="utf-8")

    print(f"\nExtraction complete! Results saved to {output_csv}")

# Example Usage
folder_path = "path/to/your/image/folder"  # Replace with actual path
process_images_in_folder(folder_path)
```
